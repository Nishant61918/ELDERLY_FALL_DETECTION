```python
# ============================================================================
# SisFall 1D-CNN for Kaggle Notebook
# Copy each cell below into separate cells in Kaggle
# ============================================================================

# ============================================================================
# CELL 1: Fix TensorFlow/Protobuf Compatibility (Run First if Errors Occur)
# ============================================================================
# Uncomment and run if you get protobuf errors
# !pip install protobuf==3.20.3

# ============================================================================
# CELL 2: Imports
# ============================================================================
import os
import glob
import re
import math
import argparse
from io import StringIO

import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.utils import class_weight
from sklearn.metrics import classification_report, confusion_matrix

# Display function for notebook compatibility
try:
    from IPython.display import display
except Exception:
    def display(obj):
        """Fallback display for scripts"""
        try:
            if hasattr(obj, 'to_string'):
                print(obj.to_string())
            else:
                print(obj)
        except Exception:
            print(obj)

print("TensorFlow version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))

# ============================================================================
# CELL 3: Check Dataset Location (UPDATE PATHS AFTER RUNNING THIS)
# ============================================================================
# Run this to see your dataset structure
print("=== Contents of /kaggle/input/ ===")
!ls /kaggle/input/

# If you have subdirectories, check them too
print("\n=== Checking for subdirectories ===")
for item in os.listdir('/kaggle/input/'):
    path = os.path.join('/kaggle/input/', item)
    if os.path.isdir(path):
        print(f"\nContents of {item}:")
        !ls /kaggle/input/{item}

# ============================================================================
# CELL 4: Configuration - UPDATE THESE PATHS!
# ============================================================================
# IMPORTANT: Update these paths based on the output from Cell 3

# Example sample file path (dry-run) - UPDATE THIS!
SAMPLE_FILE = "/kaggle/input/sis-fall-original-dataset/SisFall_dataset/SA01/D01_SA01_R01.txt"

# Dataset root for full training - UPDATE THIS!
DATA_ROOT = "/kaggle/input/sis-fall-original-dataset/SisFall_dataset"

# Which accelerometer to use: 1 => ADXL345 (acc1), 2 => MMA8451Q (acc2)
ACCEL_CHOICE = 1

# Windowing / sampling parameters
SAMPLE_RATE = 200            # Hz (SisFall)
WINDOW_SECONDS = 2.0         # seconds
STEP_FRAC = 0.5              # overlap fraction
WINDOW_SIZE = int(SAMPLE_RATE * WINDOW_SECONDS)
STEP_SIZE = max(1, int(WINDOW_SIZE * STEP_FRAC))

# Training hyperparameters
BATCH_SIZE = 64
EPOCHS = 70
GYRO_UNITS = 'rad'          # 'rad' or 'deg' for display/training units
FALL_PATTERN = None         # regex to match fall filenames, default uses '^F' (case-insensitive)

# Verify paths exist
print(f"Sample file exists: {os.path.exists(SAMPLE_FILE)}")
print(f"Data root exists: {os.path.exists(DATA_ROOT)}")
if not os.path.exists(SAMPLE_FILE):
    print(" WARNING: Sample file not found! Update SAMPLE_FILE path above.")
if not os.path.exists(DATA_ROOT):
    print(" WARNING: Data root not found! Update DATA_ROOT path above.")

# ============================================================================
# CELL 5: Physical Constants and Conversion Factors
# ============================================================================
# Constants from SisFall README (sensor specifications)
ADXL_RANGE_G = 16.0
ADXL_RESOLUTION = 13
ADXL_G_PER_BIT = (2.0 * ADXL_RANGE_G) / (2 ** ADXL_RESOLUTION)  # g / bit

MMA_RANGE_G = 8.0
MMA_RESOLUTION = 14
MMA_G_PER_BIT = (2.0 * MMA_RANGE_G) / (2 ** MMA_RESOLUTION)      # g / bit

ITG_RANGE_DEG = 2000.0
ITG_RESOLUTION = 16
ITG_DEG_PER_BIT = (2.0 * ITG_RANGE_DEG) / (2 ** ITG_RESOLUTION)  # deg/s per bit

G0 = 9.80665  # m/s^2 per g (standard gravity)

print("Conversion factors initialized:")
print(f"  ADXL345: {ADXL_G_PER_BIT:.6f} g/bit")
print(f"  MMA8451Q: {MMA_G_PER_BIT:.6f} g/bit")
print(f"  ITG3200: {ITG_DEG_PER_BIT:.6f} deg/s per bit")

# ============================================================================
# CELL 6: Text Parsing Helper Functions
# ============================================================================
number_re = re.compile(r'[-+]?\d*\.?\d+(?:[eE][-+]?\d+)?')

def parse_numeric_lines(text):
    """
    Parse text where each meaningful line contains numeric values separated by commas/spaces,
    optionally ending with ';'. Returns a DataFrame of floats (cols named c0..cN).
    """
    lines = []
    for raw_line in text.replace(';', '\n').splitlines():
        line = raw_line.strip()
        if not line:
            continue
        nums = number_re.findall(line)
        if nums:
            lines.append([float(x) for x in nums])
        else:
            # skip non-numeric lines (headers)
            continue
    if not lines:
        return pd.DataFrame()
    lengths = [len(r) for r in lines]
    if len(set(lengths)) != 1:
        raise ValueError(f"Inconsistent numeric columns across lines: {set(lengths)}")
    arr = np.array(lines, dtype=float)
    cols = [f'c{i}' for i in range(arr.shape[1])]
    return pd.DataFrame(arr, columns=cols)

def read_text_or_csv_file(path):
    """
    Read a file; detect header lines with letters -> try pandas.read_csv, otherwise parse numerics.
    """
    with open(path, 'r', encoding='utf-8', errors='ignore') as f:
        content = f.read()
    # examine first meaningful line for letters
    for ln in content.splitlines():
        ln = ln.strip()
        if not ln:
            continue
        if re.search(r'[A-Za-z]', ln):
            # try pandas.read_csv to handle headers
            try:
                df = pd.read_csv(StringIO(content))
                return df
            except Exception:
                break
        else:
            break
    return parse_numeric_lines(content)

print("Text parsing functions defined ")

# ============================================================================
# CELL 7: Sensor Data Conversion Functions
# ============================================================================
def convert_acc_adxl_to_ms2(arr_bits):
    """Convert ADXL345 accelerometer bits to m/s^2"""
    return arr_bits * ADXL_G_PER_BIT * G0

def convert_acc_mma_to_ms2(arr_bits):
    """Convert MMA8451Q accelerometer bits to m/s^2"""
    return arr_bits * MMA_G_PER_BIT * G0

def convert_gyro_itg_to_deg_s(arr_bits):
    """Convert ITG3200 gyroscope bits to deg/s"""
    return arr_bits * ITG_DEG_PER_BIT

def degs_to_rads(arr_degs):
    """Convert degrees to radians"""
    return arr_degs * (math.pi / 180.0)

print("Conversion functions defined ✓")

# ============================================================================
# CELL 8: Map and Convert Raw Data to Sensor DataFrame
# ============================================================================
def map_and_convert(df, accel_choice=1, gyro_units='rad'):
    """
    Convert raw sensor bits to physical units (m/s^2 for acceleration, rad/s or deg/s for gyro).
    
    Args:
        df: Raw dataframe from file
        accel_choice: 1 for ADXL345, 2 for MMA8451Q
        gyro_units: 'rad' or 'deg' for gyroscope output
    
    Returns:
        DataFrame with columns: ax, ay, az, gx, gy, gz (and label if present)
    """
    cols = df.columns.tolist()
    lower = [c.lower() for c in cols]
    name_map = {c.lower(): c for c in cols}

    expected_keys = ['acc1_x','acc1_y','acc1_z','acc2_x','acc2_y','acc2_z','gyro_x','gyro_y','gyro_z']
    if all(k in lower for k in expected_keys):
        # Named columns detected
        def g(k): return name_map[k]
        acc_prefix = 'acc1' if accel_choice == 1 else 'acc2'
        acc_x = df[g(f'{acc_prefix}_x')].astype(float).values
        acc_y = df[g(f'{acc_prefix}_y')].astype(float).values
        acc_z = df[g(f'{acc_prefix}_z')].astype(float).values
        if accel_choice == 1:
            ax = convert_acc_adxl_to_ms2(acc_x)
            ay = convert_acc_adxl_to_ms2(acc_y)
            az = convert_acc_adxl_to_ms2(acc_z)
        else:
            ax = convert_acc_mma_to_ms2(acc_x)
            ay = convert_acc_mma_to_ms2(acc_y)
            az = convert_acc_mma_to_ms2(acc_z)
        gx_bits = df[g('gyro_x')].astype(float).values
        gy_bits = df[g('gyro_y')].astype(float).values
        gz_bits = df[g('gyro_z')].astype(float).values
        gx = convert_gyro_itg_to_deg_s(gx_bits)
        gy = convert_gyro_itg_to_deg_s(gy_bits)
        gz = convert_gyro_itg_to_deg_s(gz_bits)
    else:
        # Numeric-only fallback (SisFall ordering)
        if df.shape[1] < 9:
            raise ValueError("Expecting at least 9 numeric columns per SisFall README")
        acc_offset = 0 if accel_choice == 1 else 3
        ax_bits = df.iloc[:, acc_offset + 0].astype(float).values
        ay_bits = df.iloc[:, acc_offset + 1].astype(float).values
        az_bits = df.iloc[:, acc_offset + 2].astype(float).values
        gx_bits = df.iloc[:, 6].astype(float).values
        gy_bits = df.iloc[:, 7].astype(float).values
        gz_bits = df.iloc[:, 8].astype(float).values
        if accel_choice == 1:
            ax = convert_acc_adxl_to_ms2(ax_bits)
            ay = convert_acc_adxl_to_ms2(ay_bits)
            az = convert_acc_adxl_to_ms2(az_bits)
        else:
            ax = convert_acc_mma_to_ms2(ax_bits)
            ay = convert_acc_mma_to_ms2(ay_bits)
            az = convert_acc_mma_to_ms2(az_bits)
        gx = convert_gyro_itg_to_deg_s(gx_bits)
        gy = convert_gyro_itg_to_deg_s(gy_bits)
        gz = convert_gyro_itg_to_deg_s(gz_bits)

    if gyro_units == 'rad':
        gx = degs_to_rads(gx)
        gy = degs_to_rads(gy)
        gz = degs_to_rads(gz)

    sensor_df = pd.DataFrame({
        'ax': ax.astype(float),
        'ay': ay.astype(float),
        'az': az.astype(float),
        'gx': gx.astype(float),
        'gy': gy.astype(float),
        'gz': gz.astype(float),
    })
    # propagate label column if present
    if 'label' in df.columns:
        try:
            sensor_df['label'] = df['label'].astype(int).values
        except Exception:
            pass
    return sensor_df

print("Data mapping function defined ")

# ============================================================================
# CELL 9: Windowing and Normalization Functions
# ============================================================================
def sliding_windows_for_series(series_df, sensor_columns, window_size, step, file_label=None):
    """
    Create sliding windows from time series data.
    
    Args:
        series_df: DataFrame with sensor data
        sensor_columns: List of column names to include
        window_size: Number of samples per window
        step: Step size between windows
        file_label: Optional label for entire file
    
    Returns:
        X: (n_windows, window_size, n_channels) array
        y: (n_windows,) label array
    """
    sensors = series_df[sensor_columns].values
    if 'label' in series_df.columns:
        labels = series_df['label'].values
    elif file_label is not None:
        labels = np.full(len(series_df), file_label, dtype=int)
    else:
        labels = np.zeros(len(series_df), dtype=int)

    n_samples, n_channels = sensors.shape
    windows = []
    window_labels = []
    if n_samples < window_size:
        # return empty arrays
        return np.zeros((0, window_size, n_channels), dtype=np.float32), np.zeros((0,), dtype=int)
    for start in range(0, n_samples - window_size + 1, step):
        end = start + window_size
        win = sensors[start:end]
        lab = labels[start:end]
        vals, counts = np.unique(lab, return_counts=True)
        majority_label = vals[np.argmax(counts)]
        windows.append(win)
        window_labels.append(majority_label)
    if len(windows) == 0:
        return np.zeros((0, window_size, n_channels), dtype=np.float32), np.zeros((0,), dtype=int)
    X = np.stack(windows).astype(np.float32)
    y = np.array(window_labels, dtype=int)
    return X, y

def normalize_windows(X_train, X_val, method='standard'):
    """
    Normalize window data using training statistics.
    
    Args:
        X_train: Training windows
        X_val: Validation windows
        method: 'standard' (z-score) or 'minmax'
    
    Returns:
        Normalized X_train, X_val
    """
    if method == 'standard':
        mean = X_train.mean(axis=(0,1), keepdims=True)
        std = X_train.std(axis=(0,1), keepdims=True) + 1e-8
        X_train = (X_train - mean) / std
        X_val = (X_val - mean) / std
    elif method == 'minmax':
        minv = X_train.min(axis=(0,1), keepdims=True)
        maxv = X_train.max(axis=(0,1), keepdims=True)
        X_train = (X_train - minv) / (maxv - minv + 1e-8)
        X_val = (X_val - minv) / (maxv - minv + 1e-8)
    return X_train, X_val

print("Windowing and normalization functions defined ")

# ============================================================================
# CELL 10: 1D-CNN Model Architecture
# ============================================================================
def build_1dcnn(input_shape, n_classes=1, dropout=0.4):
    """
    Build 1D-CNN model for fall detection.
    
    Args:
        input_shape: (timesteps, channels)
        n_classes: 1 for binary, >1 for multiclass
        dropout: Dropout rate
    
    Returns:
        Compiled Keras model
    """
    inputs = tf.keras.Input(shape=input_shape)
    x = inputs
    
    # Conv Block 1
    x = tf.keras.layers.Conv1D(32, kernel_size=7, padding='same', activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.MaxPool1D(pool_size=2)(x)
    
    # Conv Block 2
    x = tf.keras.layers.Conv1D(64, kernel_size=5, padding='same', activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.MaxPool1D(pool_size=2)(x)
    
    # Conv Block 3
    x = tf.keras.layers.Conv1D(128, kernel_size=3, padding='same', activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.GlobalAveragePooling1D()(x)
    
    # Dense layers
    x = tf.keras.layers.Dropout(dropout)(x)
    
    if n_classes == 1:
        outputs = tf.keras.layers.Dense(1, activation='sigmoid')(x)
        loss = 'binary_crossentropy'
        metrics = ['accuracy', tf.keras.metrics.AUC(name='auc')]
    else:
        outputs = tf.keras.layers.Dense(n_classes, activation='softmax')(x)
        loss = 'sparse_categorical_crossentropy'
        metrics = ['sparse_categorical_accuracy']
    
    model = tf.keras.Model(inputs, outputs)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss=loss, 
        metrics=metrics
    )
    return model

print("Model architecture function defined ")

# ============================================================================
# CELL 11: DRY RUN - Test on Single File
# ============================================================================
print("="*70)
print("DRY RUN - Testing on single file")
print("="*70)
print(f"Sample file: {SAMPLE_FILE}")

# Parse raw file
df_raw = read_text_or_csv_file(SAMPLE_FILE)
print(f"\n✓ Raw parsed shape: {df_raw.shape}")
print("\nFirst 6 rows of raw data:")
display(df_raw.head(6))

# Convert to sensor units
df_sensors = map_and_convert(df_raw, accel_choice=ACCEL_CHOICE, gyro_units=GYRO_UNITS)
print(f"\n Converted sensor dataframe shape: {df_sensors.shape}")
print("\nFirst 6 rows of converted data:")
display(df_sensors.head(6))

# Create windows
demo_window_size = WINDOW_SIZE
demo_step = STEP_SIZE
if len(df_sensors) < demo_window_size:
    print(f"\n Sample length {len(df_sensors)} < window size {demo_window_size}")
    print(f"   Reducing window to {len(df_sensors)} for demo.")
    demo_window_size = len(df_sensors)
    demo_step = max(1, demo_window_size // 2)

sensor_columns = ['ax','ay','az','gx','gy','gz']
X, y = sliding_windows_for_series(df_sensors, sensor_columns, demo_window_size, demo_step, file_label=None)
print(f"\n Windows shape (X): {X.shape}")
print(f" Labels shape (y): {y.shape}")

# Build and show model
if X.shape[0] > 0:
    print("\n" + "="*70)
    print("MODEL ARCHITECTURE")
    print("="*70)
    model = build_1dcnn(input_shape=X.shape[1:], n_classes=1)
    model.summary()
    print("\n Model created successfully!")
else:
    print("\n No windows created (sample too short)")

# ============================================================================
# CELL 12: Visualize First Window (Optional)
# ============================================================================
import matplotlib.pyplot as plt

if X.shape[0] > 0:
    print("="*70)
    print("VISUALIZATION - First Window")
    print("="*70)
    
    win0 = X[0]
    t = np.arange(win0.shape[0]) / SAMPLE_RATE
    
    fig, axs = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
    
    # Accelerometer plot
    axs[0].plot(t, win0[:,0], label='ax', linewidth=1.5)
    axs[0].plot(t, win0[:,1], label='ay', linewidth=1.5)
    axs[0].plot(t, win0[:,2], label='az', linewidth=1.5)
    axs[0].set_title('Accelerometer Data', fontsize=14, fontweight='bold')
    axs[0].set_ylabel('Acceleration (m/s²)', fontsize=12)
    axs[0].legend(loc='upper right')
    axs[0].grid(True, alpha=0.3)
    
    # Gyroscope plot
    axs[1].plot(t, win0[:,3], label='gx', linewidth=1.5)
    axs[1].plot(t, win0[:,4], label='gy', linewidth=1.5)
    axs[1].plot(t, win0[:,5], label='gz', linewidth=1.5)
    axs[1].set_title(f'Gyroscope Data', fontsize=14, fontweight='bold')
    axs[1].set_ylabel(f'Angular Velocity ({GYRO_UNITS}/s)', fontsize=12)
    axs[1].set_xlabel('Time (s)', fontsize=12)
    axs[1].legend(loc='upper right')
    axs[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(" Visualization complete!")
else:
    print(" Skipping visualization (no windows available)")

# ============================================================================
# CELL 13: Data Collection Function for Full Training
# ============================================================================
def gather_windows_from_root(root_dir, accel_choice, window_size, step, gyro_units='rad', fall_pattern=None):
    """
    Collect all windows from dataset directory.
    
    Args:
        root_dir: Root directory of dataset
        accel_choice: 1 or 2
        window_size: Samples per window
        step: Step between windows
        gyro_units: 'rad' or 'deg'
        fall_pattern: Regex pattern to identify fall files
    
    Returns:
        X_all: All windows
        y_all: All labels
    """
    exts = ('*.txt', '*.dat', '*.csv')
    files = []
    for e in exts:
        files.extend(glob.glob(os.path.join(root_dir, '**', e), recursive=True))
    
    if not files:
        raise FileNotFoundError(f"No files found under {root_dir}")
    
    print(f"Found {len(files)} files")
    
    X_list = []
    y_list = []
    
    if fall_pattern is None:
        fall_re = re.compile(r'^[Ff]')
    else:
        fall_re = re.compile(fall_pattern)
    
    skipped = 0
    processed = 0
    
    for fpath in sorted(files):
        basename = os.path.basename(fpath)
        try:
            df_raw = read_text_or_csv_file(fpath)
            if df_raw.empty:
                skipped += 1
                continue
            
            df_sensors = map_and_convert(df_raw, accel_choice=accel_choice, gyro_units=gyro_units)
            file_label = 1 if fall_re.search(basename) else 0
            
            sensor_cols = ['ax','ay','az','gx','gy','gz']
            Xf, yf = sliding_windows_for_series(df_sensors, sensor_cols, window_size, step, file_label=file_label)
            
            if Xf.shape[0] == 0:
                skipped += 1
                continue
            
            X_list.append(Xf)
            y_list.append(yf)
            processed += 1
            
            if processed % 100 == 0:
                print(f"  Processed {processed} files...")
                
        except Exception as e:
            print(f"  Error processing {basename}: {e}")
            skipped += 1
            continue
    
    if not X_list:
        raise ValueError("No windows collected from dataset. Check data and window parameters.")
    
    X_all = np.concatenate(X_list, axis=0)
    y_all = np.concatenate(y_list, axis=0)
    
    print(f"\n✓ Successfully processed {processed} files")
    print(f" Skipped {skipped} files")
    
    return X_all, y_all

print("Data collection function defined ")

# ============================================================================
# CELL 14: FULL TRAINING PIPELINE (Set RUN_TRAIN=True to execute)
# ============================================================================
#  WARNING: This cell requires significant RAM and GPU resources!
# Set RUN_TRAIN = True when ready to train on full dataset

RUN_TRAIN = True  # Change to True to start training

if RUN_TRAIN:
    print("="*70)
    print("FULL TRAINING PIPELINE - STARTING")
    print("="*70)
    
    # Collect all windows
    print("\n1️ Collecting windows from dataset (this may take a while)...")
    X_all, y_all = gather_windows_from_root(
        DATA_ROOT, 
        accel_choice=ACCEL_CHOICE,
        window_size=WINDOW_SIZE, 
        step=STEP_SIZE,
        gyro_units=GYRO_UNITS, 
        fall_pattern=FALL_PATTERN
    )
    
    print(f"\n Total windows collected: {X_all.shape}")
    print(f" Label distribution: {np.bincount(y_all)}")
    print(f"  - Class 0 (Non-fall): {np.sum(y_all==0)} windows")
    print(f"  - Class 1 (Fall): {np.sum(y_all==1)} windows")
    
    if np.unique(y_all).size == 1:
        print("\n WARNING: Only one label found!")
        print("   Check fall_pattern and filenames.")
    else:
        # Split data
        print("\n2️ Splitting data into train/validation sets...")
        X_train, X_val, y_train, y_val = train_test_split(
            X_all, y_all, 
            test_size=0.2, 
            stratify=y_all, 
            random_state=42
        )
        print(f" Training set: {X_train.shape}")
        print(f" Validation set: {X_val.shape}")

        mean = X_train.mean(axis=(0,1))
        std  = X_train.std(axis=(0,1))
        
        print("\n" + "="*70)
        print("RAW NORMALIZATION VALUES FOR ESP32")
        print("="*70)
        
        print("RAW_CHANNEL_MEAN =", mean.tolist())
        print("RAW_CHANNEL_STD =", std.tolist())

        
        # Normalize
        print("\n3️ Normalizing data...")
        X_train, X_val = normalize_windows(X_train, X_val, method='standard')
        print(" Normalization complete")
        
        # Calculate class weights
        print("\n4️ Computing class weights...")
        cw = class_weight.compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
        class_weights = {int(cls): float(weight) for cls, weight in zip(np.unique(y_train), cw)}
        print(f" Class weights: {class_weights}")
        
        # Build model
        print("\n5️ Building model...")
        n_classes = 1 if len(np.unique(y_train)) == 2 else len(np.unique(y_train))
        model = build_1dcnn(input_shape=X_train.shape[1:], n_classes=n_classes)
        print(" Model built")
        model.summary()
        
        # Create datasets
        print("\n6️ Creating TensorFlow datasets...")
        train_ds = tf.data.Dataset.from_tensor_slices((X_train, y_train))\
            .shuffle(len(X_train))\
            .batch(BATCH_SIZE)\
            .prefetch(tf.data.AUTOTUNE)
        
        val_ds = tf.data.Dataset.from_tensor_slices((X_val, y_val))\
            .batch(BATCH_SIZE)\
            .prefetch(tf.data.AUTOTUNE)
        print(" Datasets created")
        
        # Setup callbacks
        callbacks = [
            tf.keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss', 
                factor=0.5, 
                patience=4, 
                min_lr=1e-6,
                verbose=1
            ),
            tf.keras.callbacks.EarlyStopping(
                monitor='val_loss', 
                patience=10, 
                restore_best_weights=True,
                verbose=1
            ),
            tf.keras.callbacks.ModelCheckpoint(
                'best_model.h5',
                monitor='val_loss',
                save_best_only=True,
                verbose=1
            )
        ]
        
        # Train model
        print("\n7️ Training model...")
        print("="*70)
        history = model.fit(
            train_ds, 
            validation_data=val_ds, 
            epochs=EPOCHS, 
            class_weight=class_weights, 
            callbacks=callbacks
        )
        print("="*70)
        print(" Training complete!")
        
        # Evaluate
        print("\n8️ Evaluating on validation set...")
        y_pred_prob = model.predict(X_val, batch_size=BATCH_SIZE)
        
        if n_classes == 1:
            y_pred = (y_pred_prob.ravel() >= 0.5).astype(int)
        else:
            y_pred = np.argmax(y_pred_prob, axis=1)
        
        print("\n" + "="*70)
        print("CLASSIFICATION REPORT")
        print("="*70)
        print(classification_report(y_val, y_pred, target_names=['Non-Fall', 'Fall']))
        
        print("\n" + "="*70)
        print("CONFUSION MATRIX")
        print("="*70)
        cm = confusion_matrix(y_val, y_pred)
        print(cm)
        print(f"\nTrue Negatives:  {cm[0,0]}")
        print(f"False Positives: {cm[0,1]}")
        print(f"False Negatives: {cm[1,0]}")
        print(f"True Positives:  {cm[1,1]}")
        
        # Save model
        print("\n9️ Saving model...")
        model.save('sisfall_1dcnn_model.h5')
        print(" Model saved to 'sisfall_1dcnn_model.h5'")
        
        print("\n" + "="*70)
        print("TRAINING COMPLETE! ")
        print("="*70)

else:
    print("="*70)
    print("Training not enabled. Set RUN_TRAIN=True to train.")
    print("="*70)

# ============================================================================
# CELL 15: Plot Training History (Run after training)
# ============================================================================
# Only run this cell after training is complete

try:
    if 'history' in locals():
        fig, axes = plt.subplots(1, 2, figsize=(14, 5))
        
        # Loss plot
        axes[0].plot(history.history['loss'], label='Train Loss', linewidth=2)
        axes[0].plot(history.history['val_loss'], label='Val Loss', linewidth=2)
        axes[0].set_title('Model Loss', fontsize=14, fontweight='bold')
        axes[0].set_xlabel('Epoch', fontsize=12)
        axes[0].set_ylabel('Loss', fontsize=12)
        axes[0].legend()
        axes[0].grid(True, alpha=0.3)
        
        # Accuracy plot
        acc_key = 'accuracy' if 'accuracy' in history.history else 'sparse_categorical_accuracy'
        val_acc_key = 'val_accuracy' if 'val_accuracy' in history.history else 'val_sparse_categorical_accuracy'
        axes[1].plot(history.history[acc_key], label='Train Accuracy', linewidth=2)
        axes[1].plot(history.history[val_acc_key], label='Val Accuracy', linewidth=2)
        axes[1].set_title('Model Accuracy', fontsize=14, fontweight='bold')
        axes[1].set_xlabel('Epoch', fontsize=12)
        axes[1].set_ylabel('Accuracy', fontsize=12)
        axes[1].legend()
        axes[1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
        print(" Training history plotted")
    else:
        print("No training history available. Run training first (Cell 14 with RUN_TRAIN=True)")
except Exception as e:
    print(f"Could not plot history: {e}")

# ============================================================================
# END OF NOTEBOOK
# ============================================================================
print("\n" + "="*70)
print("Setup complete! Next steps:")
print("="*70)
print("1. ✓ Run cells 1-11 to test on a single file")
print("2. ✓ Check visualization in cell 12")
print("3.  Set RUN_TRAIN=True in cell 14 to train on full dataset")
print("4.  Enable GPU in Kaggle settings for faster training")
print("="*70)
```

```python
import tensorflow as tf
import numpy as np

print("="*70)
print("STEP 1: CONVERTING MODEL TO TENSORFLOW LITE")
print("="*70)

# Load model
model = tf.keras.models.load_model('sisfall_1dcnn_model.h5')

print("\n Model loaded")
print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)

# Converter
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]

# Representative dataset
def representative_dataset():
    for i in range(min(100, X_val.shape[0])):
        yield [X_val[i:i+1].astype(np.float32)]

converter.representative_dataset = representative_dataset

# INT8 quantization
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.int8
converter.inference_output_type = tf.int8

print("\nConverting...")

tflite_model = converter.convert()

print("\n Conversion successful!")

# Save model
with open('sisfall_model.tflite', 'wb') as f:
    f.write(tflite_model)

print(" Saved as sisfall_model.tflite")

print("Model size:", len(tflite_model)/1024, "KB")
```

```python
# ============================================================================
# CELL: Test TensorFlow Lite Model
# ============================================================================
print("="*70)
print("STEP 2: TESTING TFLITE MODEL")
print("="*70)

# Load TFLite model
interpreter = tf.lite.Interpreter(model_content=tflite_model)
interpreter.allocate_tensors()

# Get input and output details
input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print("\n Model Details:")
print(f"  Input shape: {input_details[0]['shape']}")
print(f"  Input type: {input_details[0]['dtype']}")
print(f"  Output shape: {output_details[0]['shape']}")
print(f"  Output type: {output_details[0]['dtype']}")

# Get quantization parameters
input_scale = input_details[0]['quantization'][0]
input_zero_point = input_details[0]['quantization'][1]
output_scale = output_details[0]['quantization'][0]
output_zero_point = output_details[0]['quantization'][1]

print(f"\n Quantization Parameters:")
print(f"  Input scale: {input_scale}")
print(f"  Input zero point: {input_zero_point}")
print(f"  Output scale: {output_scale}")
print(f"  Output zero point: {output_zero_point}")

# Test on a few validation samples
print("\n Testing on sample data...")
num_test_samples = 100
correct = 0

for i in range(num_test_samples):
    # Get test sample
    test_sample = X_val[i:i+1].astype(np.float32)
    true_label = y_val[i]
    
    # Quantize input (convert float32 to int8)
    quantized_input = (test_sample / input_scale + input_zero_point).astype(np.int8)
    
    # Set input tensor
    interpreter.set_tensor(input_details[0]['index'], quantized_input)
    
    # Run inference
    interpreter.invoke()
    
    # Get output
    quantized_output = interpreter.get_tensor(output_details[0]['index'])
    
    # Dequantize output (convert int8 back to float32)
    output = (quantized_output.astype(np.float32) - output_zero_point) * output_scale
    
    # Make prediction
    prediction = 1 if output[0][0] >= 0.5 else 0
    
    if prediction == true_label:
        correct += 1

accuracy = (correct / num_test_samples) * 100
print(f"\n TFLite Model Accuracy on {num_test_samples} samples: {accuracy:.2f}%")

if accuracy > 85:
    print("   Model conversion successful! Ready for ESP32!")
elif accuracy > 75:
    print("   Slight accuracy drop - acceptable for deployment")
else:
    print("   Significant accuracy loss - check quantization settings")

# Save quantization parameters (you'll need these in ESP32 code!)
print(f"\n IMPORTANT - Save these values for ESP32 code:")
print(f"   INPUT_SCALE = {input_scale}f;")
print(f"   INPUT_ZERO_POINT = {input_zero_point};")
print(f"   OUTPUT_SCALE = {output_scale}f;")
print(f"   OUTPUT_ZERO_POINT = {output_zero_point};")

print("="*70)
```

```python
# ============================================================================
# CELL: Convert TFLite to C Header File
# ============================================================================
print("="*70)
print("STEP 3: CONVERTING TO C ARRAY")
print("="*70)

# Use xxd to convert binary to C array
import subprocess

print("\n Converting sisfall_model.tflite to C array...")

# Run xxd command
result = subprocess.run(
    ['xxd', '-i', 'sisfall_model.tflite'],
    capture_output=True,
    text=True
)

if result.returncode == 0:
    # Get the output
    c_array = result.stdout
    
    # Modify variable name to be more descriptive
    c_array = c_array.replace('sisfall_model_tflite', 'fall_detection_model')
    
    # Add header guards and includes
    header_content = f"""#ifndef FALL_DETECTION_MODEL_H
#define FALL_DETECTION_MODEL_H

// Fall Detection Model - TensorFlow Lite
// Generated from sisfall_1dcnn_model.h5
// Model Size: {len(tflite_model)} bytes

{c_array}

#endif  // FALL_DETECTION_MODEL_H
"""
    
    # Save as .h file
    with open('fall_detection_model.h', 'w') as f:
        f.write(header_content)
    
    print(" Created fall_detection_model.h")
    print(f"  Array name: fall_detection_model")
    print(f"  Array length: fall_detection_model_len = {len(tflite_model)}")
    
else:
    print(" xxd command failed. Using Python alternative...")
    
    # Python alternative
    with open('fall_detection_model.h', 'w') as f:
        f.write('#ifndef FALL_DETECTION_MODEL_H\n')
        f.write('#define FALL_DETECTION_MODEL_H\n\n')
        f.write('// Fall Detection Model - TensorFlow Lite\n')
        f.write(f'// Model Size: {len(tflite_model)} bytes\n\n')
        f.write('const unsigned char fall_detection_model[] = {\n  ')
        
        # Write bytes in groups of 12
        for i, byte in enumerate(tflite_model):
            f.write(f'0x{byte:02x}')
            if i < len(tflite_model) - 1:
                f.write(', ')
            if (i + 1) % 12 == 0:
                f.write('\n  ')
        
        f.write('\n};\n')
        f.write(f'const unsigned int fall_detection_model_len = {len(tflite_model)};\n\n')
        f.write('#endif  // FALL_DETECTION_MODEL_H\n')
    
    print(" Created fall_detection_model.h using Python")

print("\n Download this file and add to your Arduino project!")
print("="*70)

```
